```python
from pathlib import Path
from src.datasets.carb_adapter import load_carb_gold_tsv
import json

# Notebook runs from: E:\slde-aft\notebooks
project_root = Path.cwd().parent

# Correct CaRB gold file path
gold_path = project_root / "data" / "CaRB" / "data" / "gold" / "test.tsv"

# Load CaRB gold extractions
gold_records = load_carb_gold_tsv(gold_path)

print("Unique CaRB sentences loaded:", len(gold_records))

# Take the first 30 unique sentences
subset_sentences = list(gold_records.values())[:30]

# Save sample inside the project data folder
output_path = project_root / "data" / "carb_dev_sample.jsonl"
output_path.parent.mkdir(parents=True, exist_ok=True)

with output_path.open("w", encoding="utf-8") as f:
    for rec in subset_sentences:
        obj = {
            "sentence_id": rec.sentence_id,
            "sentence": rec.text,
            "gold_triples": [
                {
                    "subject": t.subject,
                    "predicate": t.predicate,
                    "object": t.object,
                }
                for t in rec.gold_triples
            ],
        }
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("Saved subset to:", output_path)
print("Number of sentences saved:", len(subset_sentences))
```
