```python
from pathlib import Path
from src.datasets.carb_adapter import load_carb_gold_tsv
import json

gold_path = Path("data/CaRB/data/gold/test.tsv")
gold_records = load_carb_gold_tsv(gold_path)

# Take first 30 sentences
subset_sentences = list(gold_records.values())[:30]

# Save as JSONL for easy use later
output_path = Path("data/carb_dev_sample.jsonl")
output_path.parent.mkdir(parents=True, exist_ok=True)

with output_path.open("w", encoding="utf-8") as f:
    for rec in subset_sentences:
        obj = {
            "sentence_id": rec.sentence_id,
            "sentence": rec.text,
            "gold_triples": [
                {"subject": t.subject, "predicate": t.predicate, "object": t.object}
                for t in rec.gold_triples
            ],
        }
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("Saved subset to", output_path)
print("Number of sentences:", len(subset_sentences))
```
